```python
excel_filtrado_fechas = excel[
    (excel["Year"]>= "2001") &
    (excel["Year"]<= "2026")]
print(excel_filtrado_fechas)
#Ahora tenemos el excel filtrado entre los años 2001 y 2026. Durante el último cuarto de siglo. 
```

```python
excel_filtrado_pais = excel_filtrado_fechas[
    excel_filtrado_fechas["Country"] == "USA"
]
print(excel_filtrado_pais)
```
